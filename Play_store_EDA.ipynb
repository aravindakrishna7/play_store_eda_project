# **Project Summary -**

The Google Play Store Exploratory Data Analysis (EDA) project focuses on analyzing a dataset of Android applications to identify patterns and insights related to app categories, ratings, reviews, installations, pricing, app size, content ratings, and Android version compatibility. The main objective of this project was to understand the characteristics of successful and widely distributed applications and derive meaningful business insights that can support data-driven decision-making.

Data Cleaning and Preparation

The dataset initially contained duplicate records, missing values, inconsistent data formats, and one clearly misaligned record. The data was first inspected using functions such as head(), shape, info(), describe(), and isnull().sum(). Duplicate records were identified and removed to improve data quality. A malformed row containing inconsistent values across multiple columns was also identified and removed.

Missing values were carefully analyzed rather than blindly replaced. Missing ratings were retained because replacing them with artificial values could distort the analysis. The missing app type was filled as "Free" based on the corresponding price information. Other missing values, such as current Android version information, were retained where no reliable replacement could be determined.

Several columns required data transformation. Reviews and installs were converted into numerical formats. Special characters such as +, commas, and $ were removed where necessary. App sizes were standardized into megabytes, including conversion of kilobytes into megabytes. "Varies with device" values were treated as missing because their actual size could not be reliably determined. The Last Updated column was converted into a datetime format.

Exploratory Data Analysis

A variety of visualizations were created using Matplotlib, Seaborn, and Plotly. Category distributions showed that Family, Game, and Tools contained a large number of applications. The analysis also showed that the dataset was heavily dominated by free applications, with paid applications representing a much smaller proportion.

Rating analysis showed that available app ratings were generally concentrated toward the higher end of the 1–5 scale. The relationship between reviews and installs showed a noticeable positive association, suggesting that applications with higher visibility and user adoption often receive more reviews.

App size analysis showed that most applications were relatively small, while a smaller number of applications had much larger sizes. However, app size did not show a strong relationship with installations. Similarly, price was heavily concentrated around zero, indicating the dominance of free applications in the dataset.

Category-level analysis was used to compare average ratings, average reviews, and average installations. A correlation heatmap and pair plot were also used to examine relationships among numerical variables such as ratings, reviews, installs, size, and price. The analysis highlighted that reviews and installs had the clearest positive relationship, while variables such as price and size showed weaker relationships with installs.

Additional analysis examined content ratings, Android version compatibility, and application update years. Most applications had an "Everyone" content rating, and many applications were updated during the later years represented in the dataset.

Business Recommendations

Based on the analysis, businesses should focus on improving user engagement and satisfaction, monitoring user reviews, maintaining regular application updates, optimizing application size, and carefully evaluating category competition. Pricing strategies should also be selected based on the value provided to users rather than relying solely on a paid model.

Conclusion

Overall, this project demonstrates how exploratory data analysis can transform raw application data into meaningful business insights. The analysis provides a better understanding of the Google Play Store ecosystem and demonstrates practical skills in data cleaning, transformation, visualization, statistical analysis, and business interpretation using Python, Pandas, NumPy, Matplotlib, Seaborn, and Plotly.
